 Write a program that takes an email address as input and splits it into two parts: the <br>
"username" and the "domain".
- Example Input: support@google.com
- Example Output: Username: support, Domain: google.com

```python
em=input("please enter the email").strip()
for a in em:
    if a=='@':
        x= em.index(a)
us=em[:x]
do=em[x+1:]
print(f"Username: {us} , Domain: {do}")
# ============================another answer======================
us, do = em.split("@")
print(f"Username: {us} , Domain: {do}")
```

the program ignores any elements in the list that are not numbers (like strings or
booleans).
